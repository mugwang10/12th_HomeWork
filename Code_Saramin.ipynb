```python
# 필요 라이브러리 import
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# Selenium으로 웹 드라이버를 실행
driver = webdriver.Chrome()

# url 전달을 전달합니다.
url = 'https://www.saramin.co.kr/zf_user/search?search_area=main&search_done=y&search_optional_item=n&searchType=search&searchword=%EB%8D%B0%EC%9D%B4%ED%84%B0%EB%B6%84%EC%84%9D'
driver.get(url)
time.sleep(5)

# 크롤링 및 df 저장
divs = driver.find_elements(By.CSS_SELECTOR , 'div[class^="item_recruit"]')
time.sleep(2)

site_list = []
company_list = []
recruit_list = []
details = []
details_list = []
link_list = []

for div in divs:
    details = []
    site = 'Saramin'
    company = div.find_element(By.CSS_SELECTOR, 'a[class*="track_event data_layer"]').text
    recruit = div.find_element(By.CSS_SELECTOR, 'h2.job_tit a span').text
    span_block = div.find_elements(By.CSS_SELECTOR, 'div.job_condition span')
    for span in span_block:
        details.append(span.text) 
    link = div.find_element(By.CSS_SELECTOR, 'h2.job_tit a').get_attribute("href")
    
    site_list.append(site)
    company_list.append(company)
    recruit_list.append(recruit)
    details_list.append(details)
    link_list.append(link)
driver.quit()

data = {'Site':site_list, 'Col_Company':company_list, 'Col_Recruit':recruit_list, 'Col_detail':details_list, 'Col_url':link_list}
df = pd.DataFrame(data)
df
```

```python
# csv파일로 저장

df.to_csv('./data_tmp/data_saramin.csv')
```
